# What is Llama?

Llama (short for Large Language Model Meta AI) is a family of open-source large language models, developed by MetaAI. 

**Key features:**
- Open weights: meta releases the model weights (parameters) publicly, allowing anyone to download, run, fine-tune, or build on them.
- Sizes: Models come in various parameter counts (e.g., 7B, 70B, 405B), where bigger usually means more capable but requires more compute. 
- Capabilities: Excellent for text generation, reasoning, coding, translation, and instruction-following.



**Evolution:**
- Llama 2 (2023): Improved safety and performance.
- Llama 3 (2024): Major leap with 8B, 70B, and later 405B models; better reasoning and longer context (up to 128K tokens).
- Llama 3.1 (2024): Added 405B frontier-level model, multilingual support.
- Llama 3.3 (late 2024/early 2025): Efficient 70B text-only model.

## Vision models 

Early Llama models were text-only. Vision models came later. 

- First vision model: `Llama 3.2`
- Current vision model: `Llama 4`

# Llama 3.2 

Llama 3.2, released by Meta on September 25, 2024, marked a significant milestone: it was Meta's first release of multimodal (vision-capable) models in the Llama family. It also introduced ultra-lightweight text-only models optimized for running on edge devices like phones and tablets.

All models have a 128,000 token context window (same as Llama 3.1) and support multilingual tasks in 8 languages: English, German, French, Italian, Portuguese, Hindi, Spanish, and Thai.

# Usecases of `llama-3.2-vision` model

## 01. Image Description and Follow up Questions

```python
import ollama
from pathlib import Path

# Image path
image_path = Path(r"C:\Users\sachi\OneDrive\Desktop\multimodal-ai\multimodal-ai\data\images\04_image.jpg")

# Read image as bytes
image_bytes = image_path.read_bytes()

# Start a conversation
messages = [
    {
        "role": "user",
        "content": "Describe this image in extreme detail: objects, colors, lighting, mood, and any text.",
        "images": [image_bytes]
    }
]

response = ollama.chat(
    model="llama3.2-vision",
    messages=messages
)

print("Description:\n", response["message"]["content"])

# Follow-up question (same context, no need to resend image)
messages.append({
    "role": "assistant",
    "content": response["message"]["content"]
})

messages.append({
    "role": "user",
    "content": "What emotions do the people in the image seem to be feeling? Why?"
})

response2 = ollama.chat(
    model="llama3.2-vision",
    messages=messages
)

print("\nEmotion Analysis:\n", response2["message"]["content"])
```

# Medical Imaging

### 01. X-ray imaging

```python
import ollama
from pathlib import Path

# Use project-relative path (recommended)
image_path = Path(r"C:\Users\sachi\OneDrive\Desktop\multimodal-ai\multimodal-ai\data\images\05_xray.jpg")

# Safety check
if not image_path.exists():
    raise FileNotFoundError(f"Image not found: {image_path}")

# Read image as bytes
image_bytes = image_path.read_bytes()

messages = [
    {
        "role": "user",
        "content": (
            "You are a medical imaging assistant. "
            "Analyze this medical image in detail. "
            "Describe anatomical structures, visible abnormalities, "
            "possible pathological findings, and image quality. "
            "Do NOT give a final diagnosis. "
            "Mention that this is not a medical diagnosis."
        ),
        "images": [image_bytes]
    }
]

response = ollama.chat(
    model="llama3.2-vision",
    messages=messages
)

print("=== Medical Image Analysis ===\n")
print(response["message"]["content"])
```

### Follow up question (without no image resend)

```python
messages.append({
    "role": "assistant",
    "content": response["message"]["content"]
})

messages.append({
    "role": "user",
    "content": (
        "What are the possible differential considerations based on the image? "
        "What additional tests or imaging would be recommended?"
    )
})

response2 = ollama.chat(
    model="llama3.2-vision",
    messages=messages
)

print("\n=== Clinical Reasoning ===\n")
print(response2["message"]["content"])
```
