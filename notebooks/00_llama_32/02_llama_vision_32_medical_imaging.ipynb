# Multimodal Medical Image Analysis

## Objective
To analyze medical images by combining:
- Medical imaging data
- Clinical text
- Patient metadata

This system provides **educational decision support** and does **not perform medical diagnosis**.

```python
import ollama
from pathlib import Path
import json
```

```python
# Path to the medical image (relative path recommended)
image_path = Path(r"C:\Users\sachi\OneDrive\Desktop\multimodal-ai\multimodal-ai\data\images\05_xray_male.jpg")

# Check if image exists
print("Image exists:", image_path.exists())

# Read image as bytes
image_bytes = image_path.read_bytes()
```

```python
patient_data = {
    "age": 55,
    "gender": "Male",
    "smoking_history": "Yes",
    "symptoms": ["cough", "shortness of breath", "fever"],
    "oxygen_saturation": 92
}

patient_data
```

```python
prompt = f"""
You are an AI medical imaging assistant.

TASK:
Analyze the provided medical image along with patient metadata.

PATIENT DATA:
{json.dumps(patient_data, indent=2)}

INSTRUCTIONS:
- Identify the imaging modality
- Describe anatomical structures
- Mention visible abnormalities
- Assess severity as Low / Medium / High
- Suggest further tests (NO diagnosis, NO treatment)
- Include an ethical disclaimer

OUTPUT FORMAT (JSON ONLY):
{{
  "modality": "",
  "anatomical_region": "",
  "findings": [],
  "severity": "",
  "clinical_considerations": [],
  "recommended_follow_up": [],
  "disclaimer": ""
}}
"""
```

```python
messages = [
    {
        "role": "user",
        "content": prompt,
        "images": [image_bytes]
    }
]

response = ollama.chat(
    model="llama3.2-vision",
    messages=messages
)

analysis_output = response["message"]["content"]
analysis_output
```

```python
print("=== Multimodal Medical Image Analysis ===\n")
print(analysis_output)
```

```python
messages.append({
    "role": "assistant",
    "content": analysis_output
})

messages.append({
    "role": "user",
    "content": "What additional imaging or laboratory tests would help validate these findings?"
})

followup_response = ollama.chat(
    model="llama3.2-vision",
    messages=messages
)

print("=== Follow-up Reasoning ===\n")
print(followup_response["message"]["content"])
```
